###  Recipe 1: Advanced text normalization techniques
1. Install the required libraries by running the following code:

```python
!pip install unidecode
!pip install textblob==0.15
!pip install datetime
!pip install langdetect
```

2. Import necessary libraries by running the following code:

```python
import unidecode
from textblob import TextBlob
import re #regex
from datetime import datetime
from langdetect import detect
```

3.  **Unicode normalization**: This converts all Unicode characters in the text to their closest ASCII equivalents. This is particularly useful for removing diacritics and other special characters that might cause inconsistencies in text processing, as shown in the following code:

```python
def normalize_unicode(text):
  return unidecode.unidecode(text)
```

4. **Detecting and normalizing language**: This uses langdetect and TextBlob to detect the language of the text. If the text is not in English, it translates the text to English. This step ensures that the text analysis can be performed uniformly in a single language, enhancing the accuracy of further processing, as shown in the following code:

```python
def normalize_language(text):
  blob = TextBlob(text)
  if detect(text)!= 'en':
    return str(blob.translate(from_lang=detect(text),to='en'))
  return text
```

5. **Normalizing dates**: This function searches for dates in the format dd/mm/yyyy using a regular expression, and converts them to the ISO format yyyy-mm-dd. This normalization makes date formats consistent across the dataset, facilitating easier comparison and sorting of dates, shown as follows:

```python
def normalize_dates(text):
  # This regex pattern might need adjustments for specific formats
  dates = re.findall(r'\d{2}/\d{2}/\d{4}', text)
  normalized_dates = [datetime.strptime(date, '%d/%m/%Y').strftime('%Y-%m-%d') for date in dates]
  for date in set(dates):
    text = text.replace(date, next(iter(normalized_dates)))
  return text
```

6. **Normalizing numbers**: Finds all standalone numbers and replaces them with the word number. This generalization of numeric values can be useful for text analysis where the specific numbers are less important than the presence of a numeric value, shown as follows:

```python
 def normalize_numbers(text):
  # Example normalization: convert all integers to ‘number’
  return re.sub(r'\b\d+\b', 'number', text)
```

 7. **Handling synonyms**: This function iterates through a dictionary of synonyms, replacing occurrences of each key (word) with its corresponding value (synonym). This step helps maintain consistency in terminology within the text, which is crucial for effective text analysis and processing, shown as follows:

```python
def replace_synonyms(text, synonym_dict):
# Example synonyms
  for word, synonym in synonym_dict.items():
    text = re.sub(r'\b' + word + r'\b', synonym, text)
    return text
```

The following is an example of text and normalized text output:

```python
original_text_english = "He said, 'I am flying to N.Y. on 04/07/2023 and back on 12/07/2023. I can't wait!'"
original_text_italian="Ciao, sono Mario e sono nato il 12/04/1990. Quest'anno saro' a N.Y. per visitare la città"
synonyms = {'N.Y.': 'New York', "can't": 'cannot'}
list_text=[original_text_english,original_text_italian]
# Applying normalization functions
for original_text in list_text:
  normalized_text = normalize_unicode(original_text)
  normalized_text = normalize_language(normalized_text)
  normalized_text = normalize_dates(normalized_text)
  normalized_text = normalize_numbers(normalized_text)
  normalized_text = replace_synonyms(normalized_text, synonyms)
  print("Original Text:", original_text)
  print("Normalized Text:", normalized_text)
```

3. Function to remove HTML tags: This function uses BeautifulSoup, a powerful HTML parsing library, which handles all HTML structures and converts them to plain text by extracting content, shown as follows:

```python
def remove_html_tags(text):
  soup = BeautifulSoup(text, "html.parser")
  return soup.get_text()
```

 4. Handling nested HTML tags: This step involves removing scripts, styles, and other nested elements that are not needed, ensuring a cleaner text output, shown as follows:

```python
def clean_nested_html(text):
  soup = BeautifulSoup(text, "html.parser")
  for script in soup(["script", "style"]):  # remove all script and style tags
    script.extract()
  return soup.get_text(separator=' ')
```

5.  Preserving HTML entities: This function decodes the HTML content into readable text while preserving entities like &lt;, &gt;, etc., shown as follows:

```python
def preserve_html_entities(text):
  soup = BeautifulSoup(text, "html.parser")
  return soup.decode(formatter="html")
```

 6. Adjusting spacing and cleaning text: After removing tags, this function normalizes the spacing by removing redundant white spaces, shown as follows:

```python
def adjust_spacing(text):
  clean_text = ' '.join(text.split())
  return clean_text
```

 7. Quality assurance check: This function checks if the cleaned text is shorter than the original, which it generally should be, to ensure no non-HTML content was removed, shown as follows:

```python
def quality_check(text, original_text):
  if len(text) >= len(original_text):
    return False  # indicates potential data loss or incorrect removal
  return True
```

The following is an example of text and cleaned text output:

```python
sample_html = "<html><head><title>Test</title></head><body><h1> Heading</h1><p>This is a <b>bold</b> paragraph.</p><script>console.log('ignore this');</script></body></html>"
clean_text = remove_html_tags(sample_html)
clean_text= clean_nested_html(clean_text)
clean_text=preserve_html_entities(clean_text)
clean_text = adjust_spacing(clean_text)
print("Quality check result:", quality_check(text=clean_text, original_text=sample_html))
print("Original HTML Content:", sample_html)
print("Cleaned Text:", clean_text)
```

4. Tokenizing text: Breaking the text into individual words allows for precise spell checking on a word-by-word basis.

```python
def tokenize(text):
  return text.split()
```

 5. Detecting misspelled words: This function identifies words that do not appear in the spell checker’s dictionary, shown as follows:

```python
def find_misspelled_words(text, spell):
  tokens = tokenize(text)
  misspelled = spell.unknown(tokens)
  return misspelled
```

 6. Correcting words: This function applies the spell checker’s suggested corrections back into the original text, shown as follows:

```python
def correct_words(text, spell):
  misspelled = find_misspelled_words(text, spell)
  corrected_text = text
  for word in misspelled:
    # Get the one `most likely` answer
    corrected_word = spell.correction(word)
    corrected_text = corrected_text.replace(word, corrected_word)
  return corrected_text
```

7. Applying contextual corrections: Further refines the corrections by considering the context in which each word is used, shown as follows:

```python
def apply_contextual_correction(text, spell):
  corrected_text = correct_words(text, spell)
  # Additional context-based improvements can be coded here
  return corrected_text
```

###  Recipe 4: Punctuation removal techniques
The following steps will help you understand how to perform punctuation removal in Python:

1. Import necessary libraries by running the following command:

```python
import regex as re
```

 2. Defining punctuation and special characters sets: This function uses a regular expression to match any character that is a punctuation symbol, shown as follows:

```python
def define_punctuation():
  return re.compile(r'[\p{P}]+')
```

3.  Creating removal function: This function applies the defined punctuation pattern to remove all punctuation from the provided text, shown as follows:

```python
def remove_punctuation(text, pattern):
  return pattern.sub('', text)
```

4.  Extending to Unicode special characters: This function removes all types of punctuation and symbols, including those from various Unicode character sets, shown as follows:

```python
def remove_unicode_punctuation(text):
  unicode_punctuation = re.compile(r'[\p{S}\p{P}]+', re.UNICODE)
  return unicode_punctuation.sub('', text)
```

5.  Applying removal to sample text: This comprehensive function integrates all steps to clean a given piece of text, shown as follows:

```python
def clean_text(text):
  punctuation_pattern = define_punctuation()
  text = remove_punctuation(text, punctuation_pattern)
  text = remove_unicode_punctuation(text)
  return text
```

6.  Ensuring text integrity: This function checks the cleaned text to ensure that it retains alphanumeric content and does not inadvertently remove necessary content, shown as follows:

```python
def verify_text_integrity(original, cleaned):
  if not cleaned.isalnum() and not cleaned:
    raise ValueError("Text integrity compromised: non-alphanumeric characters detected.")
  if len(cleaned)>=len(original):
    return False
  return True
```

The following is an example of text and cleaned text output:

```python
sample_text = "Hello, world! Welcome to the universe of NLP; where processing... becomes easy?"
cleaned_text = clean_text(sample_text)
print("Text integrity passed?",verify_text_integrity(sample_text,cleaned_text))
print("Original Text:", sample_text)
print("Cleaned Text:", cleaned_text)
```

 ## Dealing with case sensitivity
 A basic text preprocessing method called case normalization makes all characters in a text the same case, usually lowercase, in order to minimize unpredictability. In many NLP applications, such as text searching and indexing, where the meaning of words is independent of their case, or when preparing data for machine learning models, this step is essential. We guarantee uniformity across the content by normalizing the case, which facilitates matching and comparing strings.

 The following are the techniques for case normalization:
 * Installing and importing libraries: Loading the required Python libraries so that text manipulation can begin.

* Defining normalization functions: Creating functions that change text’s case from lower to upper.

* Applying normalization: Applying these operations to textual data in order to attain consistent case.

* Handling mixed case words: Proper nouns and acronyms require extra care, as they could need to stay in uppercase or mixed case.

* Integration with other text processing steps: Ensuring that case normalization seamlessly integrates into a larger text preprocessing pipeline.

* Quality control: Verifying that the conversion has been applied consistently throughout the text.



###  Recipe 5: Case normalization techniques
No additional library installation is required for case normalization as Python’s standard libraries are sufficient. The following steps will help you understand how to perform case normalization in Python:

 1. Defining case normalization function: This function converts all characters in the text to lowercase, reducing variability, shown as follows:

```python
def to_lowercase(text):
  return text.lower()
```

2. Handling exceptions for proper nouns and acronyms: This function allows for exceptions, such as proper nouns or acronyms, which might need to remain in their original case, shown as follows:

```python
def normalize_case(text, exceptions):
  words = text.split()
  normalized_words = [to_lowercase(word) if word.lower() not in exceptions else word for word in words]
  return ' '.join(normalized_words)
```

3.  Applying normalization to sample text: Integrates the normalization function within a larger text processing framework, optionally taking a set of exceptions, shown as follows:

```python
def apply_normalization(text, exceptions:set=set()):
  return normalize_case(text, exceptions)
```

4.  Quality control: This function checks if the text is entirely in lowercase, verifying the normalization process, shown as follows:

```python
def check_normalization(text,exceptions:set=None):
  if exceptions:
    text =str(set(text.split(' '))-exceptions)
  if text != to_lowercase(text):
    return False  # Indicates not all text is in lowercase
  return True
```

The following is an example of text and normalized text output:

```python
sample_text = "This is an EXAMPLE of Mixed CASE Text."
exceptions = {'EXAMPLE'}
normalized_text = apply_normalization(sample_text, exceptions)
print("Check normalization:", check_normalization(normalized_text,exceptions=exceptions))
print("Original Text:", sample_text)
print("Normalized Text:", normalized_text)
```

###  Recipe 6: Whitespace removal techniques
The following steps will help you understand how to perform whitespace removal in Python:

1. Install the required libraries by running the following command:
Nothing to install in this recipe

2. Import necessary libraries by running the following command:

```python
import regex as re
```

3. Defining whitespace patterns: This regular expression pattern identifies one or more whitespace characters (spaces, tabs, newlines), shown as follows:

```python
def define_whitespace_patterns():
  return re.compile(r'\s+')
```

 4. Creating whitespace removal function: This function replaces sequences of whitespace characters with a single space and removes leading and trailing spaces, shown as follows:

```python
def remove_excess_whitespace(text):
  pattern = define_whitespace_patterns()
  return pattern.sub(' ', text).strip()
```

5.  Handling special whitespace cases: Further cleans text by replacing tabs and newlines with spaces before general whitespace cleanup, shown as follows:

```python
def remove_special_whitespace(text):
  # Replace tabs and newlines explicitly if required
  text = text.replace('\t', ' ').replace('\n', ' ')
  return remove_excess_whitespace(text)
```

6.  Ensuring text integrity: Checks and ensures that no leading or trailing whitespace remains in the text, shown as follows:

```python
def verify_text_structure(text):
  if re.search(r'^\s|\s$', text):
    raise ValueError("Text structure compromised: leading or trailing whitespace detected.")
  return text
```

7.  Applying removal to sample text: Integrates all the whitespace cleaning steps into a single function for easy application, shown as follows:

```python
def apply_whitespace_normalization(text):
  cleaned_text = remove_special_whitespace(text)
  cleaned_text = verify_text_structure(cleaned_text)
  return cleaned_text
```

The following is an example text and normalized text output:

```python
sample_text = "  This  text\t has     \n    excessive   \n\n whitespace.\n  "
normalized_text = apply_whitespace_normalization(sample_text)
print("Original Text:", sample_text)
print("Normalized Text:", normalized_text)
```

###  Recipe 7: Sentence Segmentation Techniques
The following steps will help you understand how to perform sentence segmentation in Python:

1. Install the required libraries by running the following command:

```python
!pip install nltk
```

2. Import necessary libraries by running the following command:

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk.tokenize import sent_tokenize
```

3.  Segmenting text into sentences: This simple function applies NLTK’s sentence tokenizer to divide text into a list of sentences, shown as follows:

```python
def segment_sentences(text):
  return sent_tokenize(text)
```

4.  Handling special cases: This function revises the list of sentences to correct segmentation errors that occur due to special abbreviations and cases, shown as follows:

```python
def handle_special_cases(sentences):
  adjusted_sentences = []
  for sentence in sentences:
    if sentence.endswith(('Mr.', 'Mrs.', 'Dr.', 'Inc.')):  # Add more cases as needed
      next_index = sentences.index(sentence) + 1
      if next_index < len(sentences):
        combined_sentence = sentence + ' ' + sentences[next_index]
        adjusted_sentences.append(combined_sentence)
      else:
        adjusted_sentences.append(sentence)
    else:
      if not adjusted_sentences or not adjusted_sentences[-1].endswith(('Mr.', 'Mrs.', 'Dr.', 'Inc.')):
        adjusted_sentences.append(sentence)
  return adjusted_sentences
```

5.  Testing and validating sentence segmentation: Integrates segmentation and special case handling to ensure accurate sentence boundaries, shown as follows:

```python
def validate_segmentation(text):
  sentences = segment_sentences(text)
  sentences = handle_special_cases(sentences)
  return sentences
```

The following is an example of text and segmented sentences output:

```python
text_example = "Dr. Smith loves New York. He moved there in 2010. He founded an Hi-Tec company ACME Inc. which sells computers. He’s married with Mrs. Laura Bennet. Isn’t it great? Yes, he said so."
segmented_sentences = validate_segmentation(text_example)
print("Segmented Sentences:", segmented_sentences)
```

###  Recipe 8: Text vectorization techniques
The following steps will help you understand how to perform text vectorization in Python:

1. Install the required libraries by running the following command:

```python
!pip install numpy
!pip install scikit-learn
```

2. Import necessary libraries by running the following command:

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import numpy as np
```

3.  Choosing vectorization technique: This function initializes the desired vectorizer based on the user’s choice, defaulting to CountVectorizer, shown as follows:

```python
def initialize_vectorizer(vectorizer_type='count'):
  if vectorizer_type == 'tfidf':
    return TfidfVectorizer()
  else:
    return CountVectorizer()
```

4.  Tokenization and encoding text: This function tokenizes the text and encodes it into vectors using the initialized vectorizer, shown as follows:

```python
def tokenize_and_encode(texts, vectorizer):
  vectorizer.fit(texts)
  return vectorizer.transform(texts)
```

5.  Normalizing vectorized data: Normalizes the vectorized data to ensure consistent length and scale across all vectors, shown as follows:

```python
def normalize_vectors(vectors):
  norm = np.linalg.norm(vectors.toarray(), axis=1, keepdims=True)
  return vectors / norm
```

6.  Integrating vectorized data into ML models: Combines all previous steps to prepare data for input into ML models, shown as follows:

```python
def prepare_data_for_model(texts, vectorizer_type='count'):
  vectorizer = initialize_vectorizer(vectorizer_type)
  vectors = tokenize_and_encode(texts, vectorizer)
  vectors = normalize_vectors(vectors)
  print(vectorizer.get_feature_names_out())
  return vectors
```

```python
# return a dataframe
import pandas as pd
def prepare_data_for_model_df(texts, vectorizer_type='count'):
  vectorizer = initialize_vectorizer(vectorizer_type)
  vectors = tokenize_and_encode(texts, vectorizer)
  vectors = normalize_vectors(vectors)
  # Get feature names (terms)
  terms = vectorizer.get_feature_names_out()
  # Convert to DataFrame for easier handling
  df = pd.DataFrame(vectors.toarray(), columns=terms)
  return df
```

 1. Install the required libraries by running the following code:

2. Import necessary libraries by running:

```python
import nltk
import spacy
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('punkt_tab')
```

3. Develop domain-specific tokenizers by running the following command:

```python
def custom_tokenizer(text):
  # This tokenizer will need to handle specific punctuation and formats unique to the domain.
  tokens = word_tokenize(text)
  return tokens
```

4. Integrate domain-specific vocabulary by running the following:

```python
def integrate_domain_vocabulary(text, vocabulary):
# Integrate domain-specific vocabulary into the tokenizer or text processing pipeline.
  return " ".join([vocabulary.get(word, word) for word in text.split()])
```

5. Applying contextual preprocessing rules by running the following:

```python
def apply_contextual_rules(text, rules):
  # Apply domain-specific rules that adjust text based on its context.
  for pattern, replacement in rules.items():
    text = text.replace(pattern, replacement)
  return text
```

6. Applying semantic normalization:

```python
def normalize_semantics(text, synonym_dict):
  words = text.split()
  normalized_text = ' '.join([synonym_dict.get(word, word) for word in words])
  return normalized_text
```

 7. Advanced entity recognition: This function leverages a domain-specific NLP model to recognize and process entities relevant to the domain, shown as follows:

```python
def recognize_entities(text, nlp_model):
  nlp = spacy.load(nlp_model)
  doc = nlp(text)
  return [(ent.text, ent.label_) for ent in doc.ents]
```

The following is an example of text and processed text output:

```python
text_example = "Dr. Watson reviewed the patient’s history before prescribing Acetaminophen 500mg daily and 5 mg of Roxicodone every 10 hours."
domain_synonyms = {'Acetaminophen': 'Paracetamol','Roxicodone': 'Oxycodone'}
rules = {'500mg': '500 mg'}
# Applying functions
tokenized_text = custom_tokenizer(text_example)
vocabulary_integrated_text = integrate_domain_vocabulary(" ".join(tokenized_text), domain_synonyms)
contextually_preprocessed_text = apply_contextual_rules(vocabulary_integrated_text, rules)
normalized_text = normalize_semantics(contextually_preprocessed_text, domain_synonyms)
entities = recognize_entities(normalized_text, 'en_core_web_md')  # Example of a domain-adapted model

print("Normalized Text:", normalized_text)
print("Recognized Entities:", entities)
```

 ### Recipe 11: Anonymization techniques
 The following steps will help you understand how to perform anonymization in Python:



1. Install the required libraries by running the following:

```python
!pip install spacy
!pip install faker
!python -m spacy download en_core_web_sm
```

2. Import the necessary libraries by running the following:

```python
import spacy
from faker import Faker
```

3.  Identifying sensitive information: This function uses spaCy to detect entities likely to be personal information., shown as follows:

```python
def identify_sensitive_data(text):
  nlp = spacy.load('en_core_web_sm')
  doc = nlp(text)
  return {ent.text: ent.label_ for ent in doc.ents if ent.label_ in ['PERSON', 'GPE', 'DATE', 'ORG']}
```

4.  Applying anonymization techniques: This function replaces sensitive information with synthetic but plausible data using the Faker library, shown as follows:

```python
def anonymize_text(text, sensitive_info):
  faker = Faker()
  anonymized_info = {}
  for info, label in sensitive_info.items():
    if label == 'PERSON':
      anonymized_info[info] = faker.name()
    elif label == 'GPE':
      anonymized_info[info] = faker.country()
    elif label == 'DATE':
      anonymized_info[info] = faker.date()
    elif label == 'ORG':
      anonymized_info[info] = faker.company()
  for original, anon in anonymized_info.items():
    text = text.replace(original, anon)
  return text, anonymized_info
```

5.  Validating anonymization: This function checks if any sensitive information remains in the anonymized text, shown as follows:

```python
def validate_anonymization(anonymized_text, original_info, anonymized_info):
# Validation is successful if no original entity remains in the anonymized text
  for original, _ in original_info.items():
    if original in anonymized_text:
      return "Validation Failed"
  # Ensure that replacements are indeed different
  for original, anon in anonymized_info.items():
    if original == anon:
      return "Validation Failed"
  return "Validation Successful"
```

6. Logging and auditing process function is shown in the following command:

```python
def log_anonymization_process(info):
  print("Anonymization process:", info)
```

7. Handling re-identification risks for anonymize info is shown in the following code:

```python
def assess_re_identification_risk(anonymized_info):
  """Assess risks based on how distinguishable the anonymized entities are from the originals """
  risk_levels = []
  for original, anon in anonymized_info.items():
    if original.lower() == anon.lower():
      risk_levels.append('High risk')
    else:
      risk_levels.append('Low risk')
  if 'High risk' in risk_levels:
    return "High risk of re-identification"
  return "Low risk of re-identification"
```

The following is an example text and anonymized text output:

```python
sample_text = "Dr. John Doe works at the University of Springfield, is an associate professor of Computer Science and was born on June 6, 1965."
sensitive_info = identify_sensitive_data(sample_text)
anonymized_text, anonymized_info = anonymize_text(sample_text, sensitive_info)
validation = validate_anonymization(anonymized_text, sensitive_info, anonymized_info)
log_anonymization_process(validation)
risk_assessment = assess_re_identification_risk(anonymized_info)

print("Original Text:", sample_text)
print("Anonymized Text:", anonymized_text)
print("Validation:", validation)
print("Risk Assessment:", risk_assessment)
```
