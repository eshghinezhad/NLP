### <b>Tokenization</b>

In this section, we will look at several tokenization methods, such as whitespace-based, punctuation-based, rule-based, and statistical approaches. We will also learn how to effectively implement these methods in Python using the Natural Language Toolkit (NLTK) library, as well as how to apply these techniques to a sample text to observe and analyze the tokens that result. A key step in NLP is tokenization, which divides text into smaller, easier-to-manage chunks known as tokens. Words, phrases, or symbols that stand in for significant textual parts can be used as these tokens. To prepare raw text for additional processing and analysis in a variety of NLP applications, tokenization is an essential first step.

#### Tokenization techniques

Tokenization can be done in a variety of ways, based on the task’s particular requirements and the language’s complexity. Here are a few typical methods:

* Whitespace tokenization: In the most basic type of tokenization, text is divided using whitespace characters (space, tab, newline). It is assumed that whitespace is used to always separate words.

* Punctuation-based tokenization: Punctuation symbols, such as commas, periods, and semicolons are also utilized as delimiters to divide tokens, in addition to whitespace. Sentences and phrases inside text can be distinguished more accurately with this method.

* Rule-based tokenization: Tokens are identified using this method by applying a set of criteria (like regular expressions). Complex token structures like email addresses, dates, and contractions (such don’t as do and n’t) can be handled with its help.

* Statistical tokenization: Sophisticated techniques tokenize text using statistical models or ML. These models work well even when managing irregular token boundaries since they have been trained on vast datasets to comprehend the subtleties of a language.

 ### <b> Stemming</b>
This section digs into the concept of stemming, an important preprocessing step in NLP that reduces words to their base or root forms. This procedure is critical for lowering the complexity of text data and improving the performance of different NLP tasks including search, data retrieval, and information extraction. We will investigate the relevance of stemming in NLP, explain how it works, and learn about several stemming algorithms, with a special focus on the popular Porter Stemmer technique. We will use the NLTK library in Python to create stemming and test its efficacy on a variety of terms. By the conclusion of this section, you will have a basic understanding of how stemming may speed text processing and increase efficiency of NLP applications. In order to efficiently analyze and interpret textual material, stemming is a crucial step in NLP that reduces words to their most basic or root form. This method helps to normalize words and makes matching and comparison easier. It is especially helpful in search engines, text analysis, and information retrieval systems.

#### Stemming technique
Words are reduced to their essential form by stripping them off their suffixes during the stemming process. As an illustration, the terms running, runner, and ran are all alterations of the word run. Words are trimmed to this fundamental form by stemming algorithms like the Porter Stemmer. Text processing operations can be made more efficient by using this normalization technique, which can help to reduce the complexity and diversity of terms in text data.

#### Implementing stemming in Python
To implement stemming in Python, we will use the NLTK library and the Porter Stemmer algorithm.

```python
import nltk
#Importing NLTK and Porter Stemmer
from nltk.stem.porter import PorterStemmer
#Initializing the stemmer: Create an instance of the Porter Stemmer to be used for stemming operations:

stemmer = PorterStemmer()
#Stemming words: Use the stem method on a list of words to demonstrate the stemming process:
words = ["writing", "writer", "wrote"]
stemmed_words = [stemmer.stem(word) for word in words]
print(stemmed_words)
```

Handling complex words: To observe the stemmer’s behavior with more complex words, apply it to a varied list:

```python
complex_words = ["believing", "disbelief", "unbelievable"]
stemmed_complex = [stemmer.stem(word) for word in complex_words]
print(stemmed_complex)
```

Evaluating stemming performance: Evaluate the stemmer’s performance by comparing the original words with their stemmed:

```python
original_vs_stemmed = list(zip(complex_words, stemmed_complex))
for original, stemmed in original_vs_stemmed:
    print(f"Original: {original}, Stemmed: {stemmed}")
```

### <b> Lemmatization </b>

In this section, we will look at lemmatization, a technique in NLP that improves text preparation by turning words into their canonical or dictionary form, known as lemmas. Lemmatization techniques make advantage of the context of speech as well as the meaning of a word, resulting in increased accuracy in text normalization.

We will examine the distinctions between stemming and lemmatization, highlighting why lemmatization frequently produces more exact solutions for certain NLP tasks. We will show how to build lemmatization in Python using the NLTK library, with an emphasis on words and sentences, using real examples and exercises.

In addition, we will evaluate the usefulness of lemmatization for improving text processing and analysis, offering an in-depth view of its benefits in real-world applications. In NLP, the process of lemmatization is breaking down words into their most basic form, or lemma, as defined by a dictionary. In contrast to stemming, which only eliminates word ends to determine the root, lemmatization considers word context and morphological analysis, producing findings that are more precise and significant. This method is essential for text normalization, linguistic analysis, and improving text semantic understanding.

####  Lemmatization technique
Lemmatization is the process of appropriately identifying a word’s lemmata by analyzing its morphology while accounting for its part of speech. As it employs a lexicon and morphological examination of words to confirm that the root word (lemma) is a part of the language, it is a more advanced method than stemming. For example, better gets lemmatized to good based on its context and part of speech. By offering a consistent representation of words with comparable meanings, the procedure makes a substantial contribution to several NLP applications, including information retrieval, semantic analysis, and text summarization.

#### Implementing lemmatization in Python

```python
# Importing NLTK and WordNet lemmatizer:
import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer

# Initializing the lemmatizer: Create an instance of WordNetLemmatizer for lemmatizing words:
lemmatizer = WordNetLemmatizer()

# Lemmatizing words with part-of-speech tagging: Lemmatize words based on their part of speech to ensure accurate conversion to their base form:
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('punkt')
```

```python
from nltk import pos_tag, word_tokenize

sentence = "The leaves on the tree are falling rapidly due to the strong wind."
words = word_tokenize(sentence)
tagged_words = pos_tag(words)
print(words)
print(tagged_words)
from nltk.corpus import wordnet

def get_wordnet_pos(treebank_tag):
    if treebank_tag.startswith('J'):
        return wordnet.ADJ
    elif treebank_tag.startswith('V'):
         return wordnet.VERB
    elif treebank_tag.startswith('N'):
         return wordnet.NOUN
    elif treebank_tag.startswith('R'):
         return wordnet.ADV
    else:
        return wordnet.NOUN

lemmatized_words = [lemmatizer.lemmatize(word, get_wordnet_pos(tag)) for word, tag in tagged_words]
print(" ".join(lemmatized_words))
```

Lemmatizing sentences: Apply the lemmatizer to entire sentences to observe its effect on a larger scale:

```python
sentences = ["Me and my friends are playing with trading cards.", "The leaves were green and fresh."]
lemmatized_sentences = []
for sentence in sentences:
     words = word_tokenize(sentence)
     tagged_words = pos_tag(words)
     lemmatized = [lemmatizer.lemmatize(word, get_wordnet_pos(tag)) for word, tag in tagged_words]
     lemmatized_sentences.append(" ".join(lemmatized))

for sentence in lemmatized_sentences:
    print(sentence)
```

Evaluating lemmatization performance: Compare original and lemmatized texts to assess the accuracy and effectiveness of lemmatization:

```python
original_vs_lemmatized = zip(sentences, lemmatized_sentences)
for original, lemmatized in original_vs_lemmatized:
    print(f"Original: {original}, Lemmatized: {lemmatized}")
```

### <b> Term frequency</b>
we will look at term frequency (TF) and how it plays an important part in text mining and NLP. TF is a fundamental statistic that measures how frequently a word appears in a text, providing important information about the significance of terms inside certain publications. Understanding this idea is essential for tasks like information retrieval, document categorization, and topic modeling. We will lead through the process of computing TF with the Python NLTK package, showcasing a real application on textual data.

In addition, we will evaluate our results to acquire a better understanding of word frequency dynamics and their importance in evaluating and processing enormous quantities of text. A key idea in text mining and NLP is TF, which measures a word’s significance within a corpus or document. It is predicated on the idea that a word’s importance directly correlates with how often it appears in the text. Understanding the distribution and significance of terms within the material is made easier by analyzing TF, which is helpful for tasks like information retrieval, topic modelling, and keyword extraction.


#### Term frequency technique
The number of times a certain word appears in a document divided by the total number of words in that document yields the TF. By balancing the effect of document length on word count makes it possible to compare term importance fairly amongst various papers. To create the term frequency–inverse document frequency (TF-IDF) score, which assesses the relative importance of words by considering their frequency throughout a group of documents, TF is frequently combined with inverse document frequency (IDF).

####  Implementing term frequency in Python

```python
#Importing required modules
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt')

#Tokenizing the text: Convert the text into individual tokens (words) for frequency analysis:)
text = "Natural language processing allows computers to understand human language. This technology is a branch of artificial intelligence."
tokens = word_tokenize(text)

#Calculating TF: Compute the frequency of each word in the text:
from collections import Counter

word_counts = Counter(tokens)
total_words = len(tokens)
term_frequency = {word: count / total_words for word, count in word_counts.items()}

#Displaying term frequencies:
for word, freq in term_frequency.items():
    print(f"{word}: {freq:.3f}")

#Analyzing TF results: Evaluate the TF results to identify significant words and patterns in the text
most_common_words = word_counts.most_common(5)
print("Most common words and their frequencies:")
for word, count in most_common_words:
    print(f"{word}: {count} ({term_frequency[word]:.3f})")
```

### <b> Inverse document frequency</b>
As we explore the complexities of text analysis in NLP, the next section introduces the idea of IDF. IDF is a key statistic for determining the relevance of a word inside the corpus, as it considers not only its frequency but also how distinctively it appears across various sources.

This measure aids in distinguishing between common words and those that are meaningful and instructive about the contents of a document.
We will look at the theoretical basis of IDF and its importance in finding essential phrases that define a text.

Using the Python NLTK library, we will show how to perform IDF calculations and then evaluate the results to better understand the distribution and meaning of words in huge datasets. A statistical metric called IDF is employed in NLP to assess a word’s significance within a group or corpus of texts.

It is predicated on the idea that words that appear in a large number of documents provide less information than terms that appear in fewer documents. In text mining and information retrieval, IDF is frequently combined with TF to produce the TF-IDF score, a crucial indicator of a word’s importance to a document within a collection.

#### Inverse document frequency technique
* The logarithm of the total number of documents in the corpus divided by the number of documents containing the word is used to determine the IDF of a word.
* By highlighting words that are <b> distinctive </b> to a certain document, the logarithmic scale lessens the weight of phrases that appear often throughout the texts.
* The idea behind IDF is to increase the <b> discriminative strength </b> of term weighting by assigning a high value to terms that are uncommon throughout the corpus but may have important meaning in certain documents.

#### Implementing inverse document frequency in Python
To implement IDF calculation in Python, we will use the NumPy and NLTK libraries.

```python
# Importing libraries and preparing the corpus
import numpy as np
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt')
corpus = [
 "The quick brown fox jumps over the lazy dog",
 "The fox is quick and the fox is brown",
  "The dog is lazy but the brown fox is quick"
 ]

#  Tokenizing and creating a document frequency dictionary: Tokenize each document and calculate the document frequency for each word in the corpus:
from collections import defaultdict
document_freq = defaultdict(int)
for document in corpus:
    words = set(word_tokenize(document.lower()))
    for word in words:
        document_freq[word] += 1
# Calculating IDF: Compute the IDF for each word in the corpus:
total_documents = len(corpus)
idf = {word: np.log(total_documents / freq) for word, freq in document_freq.items()}

# Displaying IDF values: Print the IDF values for each word to understand their relative importance across the corpus:
for word, value in idf.items():
    print(f"{word}: {value:.3f}")

 #Analyzing and interpreting IDF results: Review the IDF scores to identify words with higher significance in the corpus:
sorted_idf = sorted(idf.items(), key=lambda item: item[1], reverse=True)
print("Words with highest IDF values:")
for word, value in sorted_idf[:5]:
    print(f"{word}: {value:.3f}")
```

####  NLP pipeline technique
To obtain the intended analytical result, an NLP pipeline normally begins with raw text input and processes it via several phases. The procedure consists of:

    Preprocessing: Text should be cleaned and normalized to get rid of noise and irregularities.

    Tokenization: It is the process of dividing a text into discrete words or units.

    POS tagging: Classifying each word according to its grammatical category.

    Named entity recognition (NER): Locating and categorizing the text’s named entities (people, places, and organizations).

    Dependency parsing: Linking words together by examining a sentence’s grammatical structure.

####  Part-of-speech tagging technique
Words are given labels using POS tagging, which considers their definition and sentence context. ### <b> Word sense disambiguation </b>

In this chapter, we turn our attention to word sense disambiguation (WSD), a critical and challenging task in NLP. WSD involves determining which sense of a word is used in a given context, particularly when the word has multiple meanings. This capability is fundamental for enhancing the understanding and interpretation of texts, as it clarifies ambiguities that can confuse both humans and machines. We will explore the importance of WSD and how it contributes to more accurate language models. Using the NLTK package in Python, we will implement WSD using the Lesk algorithm, a classical approach to this problem. By examining and analyzing the results, we will assess the effectiveness of our methods in achieving precise and contextually appropriate interpretations of words in text. When a word has more than one meaning, WSD, is a technique used in NLP to identify which sense the word is used in a phrase. This work is critical for many NLP applications, including content analysis, machine translation, and information retrieval, where accurate interpretation depends on knowing the correct meaning of words in context.

#### Word sense disambiguation technique
Using linguistic algorithms and contextual cues, WSD identifies a word’s intended meaning from the surrounding text. The problem is that many words are polysemous, meaning they have more than one meaning. The right meaning can only be inferred from the context. For instance, the term bank might describe a financial organization or the bank side of a river. WSD methods examine the context a word appears in to ascertain its precise meaning in that particular situation.

 #### Implementing word sense disambiguation in Python
To implement WSD in Python, we will use the NLTK library and the Lesk algorithm

```python
# Importing libraries and downloading resources
import nltk
from nltk.wsd import lesk
nltk.download('wordnet')
nltk.download('omw-1.4')

# Tokenizing the text: Break down the text into individual words for analysis:
#nltk.download('punkt')
from nltk.tokenize import word_tokenize

sentence = "The bank can guarantee deposits will eventually cover future tuition costs because it invests in adjustable-rate mortgage securities."
words = word_tokenize(sentence)

#Applying WSD: Use the Lesk algorithm, a classic WSD method, to predict the sense of a word in context:
word_to_disambiguate = "bank"
sense = lesk(words, word_to_disambiguate)

#Displaying the disambiguated sense:
print(sense, sense.definition())
```
