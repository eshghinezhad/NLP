 ## Semantic representation techniques
 The bag-of-words (BoW) model is the starting point of the journey; it is a simple yet powerful technique for text representation. This paradigm converts text into a set of tokens while keeping frequency constant and ignoring order, giving algorithms a basic degree of comprehension to work with. BoW is especially helpful in situations that need a simple quantitative examination of text, such document classification and initial topic identification. However, the main drawback of BoW is its simplicity: contextual information is lost, which is frequently essential for more complex language comprehension.

TF-IDF Calculation is a more sophisticated version of the BoW model that weighs terms according to their relevance, derived from their frequency within a text compared to their frequency throughout all papers. By reducing the influence of commonly used but less informative words, this technique improves the BoW and improves the feature set for tasks like search and information retrieval.

By capturing the local word order and offering a view into the contextual usage of words, N-gram modeling advances much farther. N-grams add a syntactic structural layer by taking into account tuples of consecutive words, which is very useful for statistical language modeling and speech recognition applications. However, higher computing costs and possible problems with data sparsity accompany the increasing granularity.

An even more sophisticated method is introduced by word embeddings and sentence embeddings, which map words and sentences into a continuous vector space where semantic links correspond to spatial proximity. These methods are essential for tasks like advanced sentiment analysis and machine translation, which call for a deep semantic knowledge of text. By considering the semantic similarity between words or sentences, they enable models to understand minute details.

A major development in NLP is the advent of pre-trained language models, such as BERT. These models provide contextually enriched embeddings and achieve breakthrough performance across a wide range of NLP tasks by utilizing large amounts of data and sophisticated neural network architectures. They provide unparalleled flexibility and power in NLP applications due to their fine-tuning capability, which allows them to adapt to unique tasks while maintaining a deep comprehension of linguistic nuances.

The specialized techniques, semantic role labeling and relationship extraction try to comprehend the underlying structure of sentences by determining the relationships between things inside a text. This is important for tasks like knowledge graph generation and information extraction. These techniques demonstrate the level of comprehension achievable with sophisticated NLP by looking into the syntax and semantics of language to extract structured relationships.

Various features of text representation and similarity measurement are addressed by techniques, such as cosine similarity, semantic clustering, and one-hot encoding. Although it ignores semantic complexity, one-hot encoding offers a simple technique for encoding items in a form that computers can comprehend. Similar entities are grouped together using semantic clustering, which improves information organization and retrieval. For applications such as plagiarism detectors and search engines’ ability to retrieve pertinent documents, cosine similarity provides a measure of how similar two documents are to one another.

With regard to the difficult issue of comprehending and interpreting natural language, each of these semantic representation strategies brings a unique set of tools to the table. NLP practitioners can address a wide range of applications by combining these diverse approaches, pushing the limits of what robots can comprehend and accomplish with human language. ##  Bag-of-words model
 Transforming textual input into a numerical format that can be easily comprehended and processed by computers is an important step in the field of NLP. The BoW model offers one of the most straightforward yet efficient ways to accomplish this transformation. The BoW model preserves multiplicity while ignoring grammar and word order, portraying text as a collection of individual words. Regarding tasks like document classification, this model is especially helpful because it can reveal more information about a word’s presence and frequency than its sequence of appearance.




###  BoW techniques
Here are some techniques of BoW:

* Text preprocessing: Depending on the application, the original text data is cleaned up by eliminating special characters, punctuation, and occasionally common terms (stop words).

* Tokenization: This is an important stage that defines how the words are identified; the cleaned text is divided into distinct terms or words.

* Vocabulary creation: Using the tokens, a vocabulary of distinct words is formed. Each distinct word is then given a location in the resultant vector, which becomes the feature set.

* Vectorization: Every text document or sentence is vectorized, with each dimension denoting the vocabulary and each value denoting the frequency of a word in the document.

* Normalization (optional): The raw frequency numbers may occasionally be normalized to take document length into consideration or to avoid features with higher counts from overshadowing features with lower counts.

Despite being straightforward, this approach serves as the basis for more intricate text representation strategies and is crucial to comprehending machine learning algorithms when applied to text.

##  TF-IDF
 Determining the significance of words within a corpus of documents in text analysis and NLP is crucial for many applications, including content summarization, information retrieval, and search engines. A statistical metric called TF-IDF is employed to assess a word’s significance to a document in a corpus or collection. This method works especially well in situations when word frequency varies greatly between papers.

###  TF-IDF techniques
The two parts of TF-IDF are the IDF and the TF. When combined, these elements provide a balance between a word’s frequency-based significance and its ubiquity throughout various sources. The following are the factors needed to calculate TF-IDF:
* TF: Indicates how often a word occurs in a given document. To avoid bias towards longer documents, it is commonly normalized by dividing by the total amount of words in the document.

* IDF: This quantifies a term’s significance over a collection of documents. IDF is based on the notion that a term’s importance rises inversely with its frequency across several publications.

* Combining TF and IDF: Multiplying TF and IDF yields the TF-IDF score. Important terms have a larger resultant value, which helps to offset the fact that certain words appear more frequently overall.

In addition to decreasing the weight of words that appear frequently in documents (like the, is, and at), TF-IDF also helps to boost the weight of phrases that appear essential but are less common.

## N-gram modeling
 Understanding the context in which words arise greatly improves language interpretation and analysis in the field of NLP. A simple yet effective method for collecting this linguistic context is N-gram modeling. A continuous series of n elements (words or letters) from a specific text or speech sample is called an n-gram. By taking word sequences into account, this strategy goes beyond basic word frequency models and adds order and context to the study. For applications like text prediction, language modeling, and spelling correction, n-grams are essential.

 ### N-gram modeling techniques
 N-gram models are based on the supposition that a word’s likelihood is solely determined by the number of words that came before it. This presumption preserves a meaningful level of contextuality while managing to simplify the difficult process of language modeling. Typically, n-gram modeling implementation entails the following steps:

* Text preprocessing: The text is cleaned and ready by converting it to lowercase, eliminating stopwords occasionally, and removing punctuation.

* Tokenization: A list of tokens (words) is created by splitting the cleaned text.

* N-gram generation: The token list is used to create N-grams. To do this, tuples of n consecutive words must be formed across the text.

* N-gram frequency distribution: Following the generation of n-grams, the frequency at which they occur within the corpus is determined.

* Probability estimation: Frequently, methods such as maximum likelihood estimation are used to assign probabilities to sequences based on their frequencies.

* Smoothing (optional): The probability estimates for unseen n-grams might be adjusted by using smoothing techniques, as not all conceivable n-grams will show up in the training data.

Depending on the value of n, N-gram models can differ greatly in terms of accuracy and complexity. More context is indicated by a greater n, but managing the continually increasing number of possible n-grams demands more data and processing capacity.

 ### Recipe 3: N-gram modeling
 This recipe will demonstrate how to implement a basic n-gram model using Python’s NLTK, which provides comprehensive support for building NLP models. Follow these steps:

1.  Install required libraries: Ensure that you have the NLTK library installed, which will be used to handle text processing and N-gram generation:

```python
!pip install nltk
```

2.  Import necessary modules: Import the necessary components from NLTK:

```python
import nltk
nltk.download('punkt_tab')
from nltk.util import ngrams
from nltk import FreqDist
```

3. Define sample text: Define some text to work with. This text will be converted into N-grams:

```python
text = "The quick brown fox jumps over the lazy dog. The quick brown fox is very friendly."
```

4.  Tokenize and preprocess: Tokenize the text and convert all tokens to lowercase:

```python
tokens = nltk.word_tokenize(text.lower())
```

5. Generate N-grams: Generate N-grams from tokens. Here, we choose trigrams (n=3) as an example:

```python
trigrams = list(ngrams(tokens, 3))
print ("3-grams",trigrams)
```

6.  Compute frequency distribution: Calculate the frequency distribution of the trigrams in the text:

```python
freq_dist = FreqDist(trigrams)
```

 7. Display the most common N-Grams: Print the most common N-grams and their frequencies:

```python
print("Most common trigrams:", freq_dist.most_common(5))
```

The number next to each tuple indicates how frequently that particular trigram appears in the text. Each tuple represents a trigram that was taken from the text. This result offers an overview of the most common contextual patterns found in the supplied text.

 ## Word embeddings
 A complex technique in NLP is word embeddings, which convert textual data into a numerical representation by mapping words with similar meanings to nearby points in a high-dimensional space. This method is an effective tool for many NLP applications, including text classification, sentiment analysis, and machine translation. It makes it easier to catch contextual subtleties, syntactic patterns, and semantic commonalities within the language.

The idea behind word embeddings is that words that appear in comparable settings are probably going to have similar meanings. The foundation of several embedding models, including Word2Vec, GloVe, and FastText, is this idea, which is frequently captured in the distributional hypothesis of linguistics.

 ### Word embeddings techniques
 Creating word embeddings usually entails the following crucial steps:

  * Corpus preparation: In order to examine the linguistic context of each word, a sizable corpus of text is needed.

  * Tokenization: Tokens, typically words, are used to represent the text.

  * Model selection: Selecting an embedding model (Word2Vec, GloVe, or another) that meets the requirements of the application.

  * Training: Using the text data, the model is taught to anticipate words based on their contexts, or vice versa, acquiring the embeddings in the process.

  * Dimensionality reduction: Compared to the initial vocabulary size, the word characteristics are projected into a vector space with substantially fewer dimensions.

  * Normalization and usage: These vectors can be used for downstream NLP tasks after being normalized post-training.

  Each of these stages helps to build a vector space model in which a dense vector of floating-point values represents each unique word in the corpus. Embeddings capture far greater information about word usage and meanings than more basic BoW models.

 ## Sentence embeddings
 Word embeddings offer a sophisticated semantic representation of words, which is a significant advancement in NLP. Important contextual linkages and similarities can be captured by using models such as Word2Vec. Due to this, embeddings are now essential to contemporary NLP systems, allowing computers to analyze human language in a contextually rich and computationally efficient manner.

###  Sentence embeddings techniques
Sentences are converted into numerical vectors, each of which reflects semantic information that the sentence encodes. This process creates sentence embeddings. Several crucial steps are included in this transformation process:

* Text collection and preprocessing: A sizable corpus of text is collected and subjected to preprocessing, which includes standardizing case and punctuation and sometimes eliminating less informative textual elements like stopwords.

* Vectorization strategy: Sentences are vectorized using a suitable technique. This could entail more sophisticated techniques utilizing neural networks or just a straightforward word vector average.

* Model training: To determine the most effective way to represent words as vectors, a model is trained using the corpus. This may entail supervised learning from labeled text data or unsupervised learning from the original text.

* Dimensionality reduction: To reduce the dimensionality of the sentence vectors while maintaining their semantic features, methods like principal component analysis (PCA) are used.

* Normalization: By guaranteeing that each vector fits within a comparable scale, the vectors are normalized, which improves the efficiency of subsequent activities such as similarity measurement.

* Usage in downstream tasks: The extensive semantic information these embeddings provide makes them useful as inputs for a variety of NLP tasks.

These embeddings’ primary objective is to simplify sentences while maintaining their semantic meaning, enabling computers to comprehend and analyze vast amounts of material with ease.

###  Recipe 5: Sentence embeddings
The following recipe steps show how to use the Doc2Vec model, an extension of the Word2Vec methodology that handles longer text units like sentences or paragraphs, to create simple sentence embeddings using Python’s Gensim package:

1.  Install required libraries: Ensure that you have Gensim installed, along with NLTK for basic text processing:

```python
!pip install gensim
!pip install nltk
```

2.  Import necessary modules:

```python
from gensim.models import Doc2Vec
from gensim.models.doc2vec import TaggedDocument
import nltk
nltk.download('punkt')
from nltk.tokenize import word_tokenize
```

3.  Prepare and tag the data: Prepare your corpus by tagging each document. This is essential for Doc2Vec to identify and learn from individual text units:

```python
text_corpus = [
    "Deep learning achieves great success in various AI fields.",
   "Transformers is a deep neural network that is fundamental in field of generative AI",
   "The representation of words as vectors facilitates understanding."
 ]
tagged_data = [TaggedDocument(words=word_tokenize(_d.lower()), tags=[str(i)]) for i, _d in enumerate(text_corpus)]
tagged_data
```

 4. Create the model: Instantiate and train a Doc2Vec model:

```python
model = Doc2Vec(vector_size=100, window=2, min_count=1, workers=4, epochs=40)
model.build_vocab(tagged_data)
model.train(tagged_data, total_examples=model.corpus_count, epochs=model.epochs)
```

5.  Generate sentence embeddings: Retrieve vector embeddings for each sentence:

```python
sentence_vectors = [model.dv[str(i)] for i in range(len(text_corpus))]
sentence_vectors
```

6.  Use embeddings for similarity comparison: Demonstrate the use of embedding by finding similarity between sentences:

```python
similarity = model.dv.n_similarity(['0'], ['1'])
```

```python
print("Similarity between first two sentences:", similarity)
```

A sentence from the corpus is embedded into each vector in the output, which is a dense vector of floating point values. Based on their embeddings, the similarity score shows how semantically similar the first two sentences are to one another.

## BERT
The emergence of models such BERT marks a paradigm shift in the field of NLP as it pertains to the understanding of human language by machines. Google created BERT, a pre-trained model that uses conditioning on both left and right context in all levels to pre-train deep bidirectional representations. Therefore, without requiring significant task-specific architecture modifications, the pre-trained BERT model may be refined with just one more output layer to produce state-of-the-art models for a variety of tasks, including question answering, sentiment analysis, and language inference.

 ### BERT techniques
 BERT’s transformer architecture, which powers its profound comprehension of linguistic context and nuance, is the key to its efficacy. Several essential steps are involved in using BERT:

* Pre-training: BERT is first pre-trained on a sizable corpus of text, gaining a broad grasp of context and linguistic structure.

* Tailoring: Additional task-specific layers can be added to BERT to fine-tune it for particular activities. This is merely tweaking the model to fit the specifics of the intended purpose.

* Tokenization: The WordPiece tokenization technique must be used to tokenize text in a particular way for BERT to function. This deconstructs words into smaller, more digestible chunks so that embeddings can be learned more successfully.

* Embedding extraction: To capture deep contextual links, BERT transforms these tokens into embeddings, which are subsequently processed through its layers of neural networks.

* Output layer: Various output layers, such as classification layers for sentiment analysis, can be built on top of BERT to generate predictions, depending on the task at hand.

* Inference: Using fresh input text, the refined model is subsequently utilized to infer or forecast results.

BERT’s power is in its deep contextual insights, which allow it to adapt to a wide range of NLP jobs with little change.

 ### Recipe 6: BERT
 This recipe walks you through the use of the Hugging Face transformers library, a well-liked toolkit that offers a simple API to interact with BERT and other transformer models, to create BERT for a classification assignment. The following steps will help you understand how to use BERT in Python:

##  One-hot encoding
In machine learning processes, one-hot encoding is a key technique that transforms categorical input into a numerical representation. It is extensively utilized for processing both textual and categorical data. One-hot encoding converts words into vectors for use in NLP. Each word is represented by a vector with one hot (1) element and all other elements being cold (0). This method offers a concise, although not semantically rich, representation of text data, making it especially helpful for models that need numerical input.

 ### One-hot encoding techniques
 One-hot encoding requires the following crucial steps:

* Vocabulary creation: Every term (vocabulary) that appears in the dataset is listed individually.

* Vector representation: A vector represents each word in the lexicon. This vector’s length corresponds to the vocabulary’s size.

* Encoding: Every word in its vector has its associated index set to 1, and all other locations are set to 0.

* Context ignorance: It is crucial to remember that one-hot encoding does not save word relationships or other contextual information. Every word has its own unique encoding.

* Sparsity: As the vocabulary expands, the sparse (mainly zeros) vectors produced by one-hot encoding may become computationally demanding.

Although this approach is straightforward, it serves as a foundation for more intricate representations and is employed in several neural network and algorithm preprocessing stages that demand a fixed-length input.

###  Recipe 8: One-hot encoding
This recipe will demonstrate how to implement one-hot encoding from scratch and with the use of high-level libraries in Python:

1.  Install required libraries:

```python
!pip install numpy
!pip install scikit-learn
```

2.  Import necessary modules:

```python
import numpy as np
from sklearn.preprocessing import OneHotEncoder
```

3. Define sample text:

```python
words = ['apple', 'orange', 'banana', 'apple', 'orange']
```

4. Create unique word index:

```python
word_dict = {word: idx for idx, word in enumerate(sorted(set(words)))}
```

5. Encode using custom function:

```python
def one_hot_encode(word):
  vector = np.zeros(len(word_dict))
  vector[word_dict[word]] = 1
  return vector

encoded_words = np.array([one_hot_encode(word) for word in words])
```

6. Use scikit-learn for one-hot encoding:

```python
encoder = OneHotEncoder(sparse_output=False)
encoded_words_sklearn=encoder.fit_transform(np.array(words).reshape(-1, 1))
```

7. Display the encoded vectors:


```python
print("Custom One-Hot Encoding:\n", encoded_words)
print("Scikit-Learn One-Hot Encoding:\n", encoded_words_sklearn)
```

Words in the input are represented by rows in the matrix, and the presence of a word is indicated by a 1 in each row according to its indexed position as defined by the vocabulary’s alphabetical order.
